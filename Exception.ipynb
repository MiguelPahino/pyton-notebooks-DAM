## Explicación de errores en python

---

En Python podemos usar `try`, `except`, `else` y `finally` para trabajar con errores. 


```python
sep = '-' * 45 + '\n'
print(sep + 'EXCEPTION RAISED AND CAUGHT')
try:
    x = 'spam'[99]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

Try es un método que depende de otros métodos como exception y else. Se utiliza para que ejecute un bloque de código. Luego están except y finally, except sirve para que si en el try se encuentra una excepción se ejecute, y el finally es la acción que se hace después de recorrer el try y las excepciones anteriores.

```python
print(sep + 'NO EXCEPTION RAISED')
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

En este caso no se lana ningún error debido a que no se encuentra ninguna excepción

```python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE')
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
else:
    print('else run')
finally:
    print('finally run')
print('after run')
```

Luego está else, se usa para que si no se ha alcanzado ninguna excepción haga la acción que indica el else.

```python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT')
try:
    x = 1 / 0
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

En este caso se lanza error por que no se puso ningún tipo de excepción para `ZeroDivisionError`.
