# Dictionaries

Los diccionarios son una forma para trabajar con propblemas que necesitas listas o datos que quieres relacionar.

La diferencia con las listas es que los dicionarios no buscan por posicionamiento de memoria, sino que usa un "key" para relacionar los datos entre si.

## Colecciones desordenadas de obejetos arbitrarios

Los diccionarios no guardan los diccionarios tal y como los mandamos nosotros, cambia el orden para buscar de forma más efectiva y luego usa un valor simbólico(key) para buscar.
Respectoa lo de arbitrario se refiere a que puede tener valores de varios tipos en el.

## Operaciones comunes

```python
D= dict(name="Miguel", age=40)
```

Esta acción sirve para crear diccionarios con variables

```python
D1 = dict([('name', 'Bob'), ('age', 40)])

keylist= [3,2,1]
valueslist= [1,2,3]
D2 = dict(zip(keylist, valueslist))
```

Otras formas de hacerlo

```python
D1["age"]
```
