## Asignar un string a una variable,

```python
# Make a 4-character string, and assign it to a name <= esto es un comentario en Python
S = 'Spam'
print(S)
```

## Length

```python
len(S)
```

## El primer item es S, indexando con la posicion 0
In Python, indexes are coded as offsets from the front, and so start from `0:` the first item is at index `0`, the second is at index `1`, and so on.



```python
S[0]
```

## Poniendo 1 muestra el segundo valor.

```python
S[1]
```

```python
# Si te sales de los límites
S[8]
```

## Para el último valor
En python se usa el -1 para que empiece desde atrás

```python

S[-1]
```

## El penúltimo valor de S

```python
S[-2]
```

## Slice of S

```python
# from offsets 1 through 2 (not 3)
S[1:3]
```

```python
# Everything past the first (1:len(S))
S[1:]
```

```python
# S itself hasn't changed
S
```

```python
# Everything but the last
S[0:3]
```

```python
# Same as S[0:3]
S[:3]
```

```python
# Everything but the last again, but simpler (0:-1)
S[:-1]
```

```python
# All of S as a top-level copy (0:len(S))
S[:]
```

## Concatenación y repetición

```python
S + 'eggs'
```

```python
# S is unchanged
S
```

```python
# Repetition
S * 6
```

## Polimorfismo
En Python, el operador + cambia de función según el tipo de objeto —suma números y concatena cadenas— gracias al polimorfismo y al tipado dinámico, que permiten que una misma operación funcione con distintos tipos siempre que estos ofrezcan una interfaz compatible.

## Inmutabilidad

Immutable objects no se pueden modificar.

Cada objeto de python se caracteriza por se immutable (unchangeable) o no. 


```python
S[0] = 'z'
```

## Podemos usar expresiones para variables

```python
S = 'z' + S[1:]
S
```

## Type-Specific Methods
### Find

```python
# Find the offset of a substring in S
S = 'SpamEggsSpam'
S.find('pa')
```

```python
S.find('pa', 3)
```

```python
S.find(S)
```

```python
# Si no existe el caracter => -1
S.find('z')
```

### Replace

```python
S.replace('Eggs', 'Bacon')
# S es inmutable! S = 'SpamEggsSpam'
```

### Mayúsculas

```python
S.upper()
# S es inmutable! S = 'SpamEggsSpam'
```

```python
print("S is alpha", S.isalpha())
print("S is digit", S.isdigit())
```

### Split

```python
S = 'spams-spam-eggs-spam-bacon'
S.split('-')
```

### Rstrip
Remove whitespace characters on the right side

```python
line = 'aaa\t,bbb\t,\nccccc,\tdd\n'
print(line)
```

```python
line.rstrip()
```

```python
line.lstrip()
```

```python
# Combine two operations
line.rstrip().split(',')
```

### Formatting
Las cadenas también permiten una operación avanzada de sustitución llamada formateo, que sirve para insertar valores dentro de un texto de manera flexible.

```python
# # Formatting expression
'%s, eggs, and %s' % ('spam', 'SPAM!')
```

```python
# Formatting method
'{}, eggs, and {}'.format('spam', 'SPAM!')
```

```python
S = 'SpamEggsSpam'
f'{S}'
```

```python
'{0}'.format(S)
```
